```python
#  Obtaining all categorical cols to have a look 
categorical_cols = df.select_dtypes(object).columns
categorical_cols
```

```python
#  Obtaining all numerical cols to have a look 
numerical_cols = df.select_dtypes('number').columns
numerical_cols
```

```python
#  Now fill rows with missing values as "unknown"
missing_categorical_cols = [
    "Teacher_Quality",
    "Distance_from_Home",
    "Parental_Education_Level"
]
df[missing_categorical_cols] = df[missing_categorical_cols].fillna("Unkown")
```

```python
#  Zero missing values, as converted to "unknown"
df[missing_categorical_cols].isnull().sum()
```

```python
# Now lets define the model into x(input) and y (output)
x = df.drop("Exam_Score", axis = 1)
y = df["Exam_Score"]
```

```python
# ==============================================================================================================================================
#                                               TRAIN - TEST - SPLIT
# ==============================================================================================================================================

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size= 0.2, random_state=42)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
# ==============================================================================================================================================
#                                               ONE - HOT ENCODING
# ==============================================================================================================================================

# handle_unknown = ignore = "Agar test data mein koi nayi category mile, error mat dena. Usko encoded representation mein ignore/zero-handle kar dena."
encoder = OneHotEncoder(handle_unknown= "ignore")
```

```python
#  See on which data you're going to do one hot encoding
x_train[categorical_cols]
```

```python
# Encoder mein fit kardiya
encoder.fit(x_train[categorical_cols])
```

```python
# Now transform
x_train_encoded = encoder.transform(x_train[categorical_cols])
```

```python
x_train_encoded.shape
```

```python
# 13 categorical columns → 37 columns aaye.
# Har categorical column mein alag alag categories hain such as some have 2 some have 3 , as private, international , govt, etc
# sab categories combine hokar 37 encoded columns bane.
```

```python
x_test_encoded = encoder.transform(x_test[categorical_cols])
```

```python
x_test_encoded.shape
```

### Ab humein 6 numerical features + 37 encoded categorical features combine karne hain:

```python
numerical_cols = x.select_dtypes('number').columns
```

```python
numerical_cols
```

#### But one small issue first: X_train_encoded and X_test_encoded are currently sparse matrices, while our numerical data is a Pandas DataFrame.
#### So before combining, we'll convert the encoded matrices into normal NumPy arrays.

```python
x_train_encoded = x_train_encoded.toarray()
```

```python
x_test_encoded = x_test_encoded.toarray()
```

```python
X_train_numerical = x_train[numerical_cols].to_numpy()
X_test_numerical = x_test[numerical_cols].to_numpy()
```

```python
import numpy as np

x_train_final = np.hstack((X_train_numerical, x_train_encoded))
x_test_final = np.hstack((X_test_numerical, x_test_encoded))
```

```python
x_train_final.shape
x_test_final.shape
```

```python
x_train_final
```

###                                                       MODEL TRAINING

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train_final, y_train)
```

### PREDICTION

```python
y_pred = model.predict(x_test_final)
```

#### Model kitna accurate hai? Iske liye hum y_pred ko y_test ke saath compare karenge.

### EVALUATION

```python
# ============================================================  MAE  ============================================================
# MAE ka simple meaning:
# Model ki predicted value actual target value se average mein kitni door hai.

# Formula:
# \[
# MAE = \frac{1}{n}\sum |Actual - Predicted|
# \]

from sklearn.metrics import mean_absolute_error
mae = mean_absolute_error(y_test, y_pred)
print("MAE", mae)
```

```python
# =====================================================   R^2 (R-squared)  ==============================================================
# R² batata hai ki model target (Exam_Score) ke variation ko kitna explain kar paa raha hai.

from sklearn.metrics import r2_score
r2 = r2_score(y_test, y_pred)
print("R2 score:", r2)
```

```python
from sklearn.metrics import root_mean_squared_error
rmse = root_mean_squared_error(y_test, y_pred)
print("RMSE", rmse)
```

### RMSE = root(MSE)

### MAE  = 0.4499
### RMSE = 1.8033
### R²   = 0.7699

```python
# RMSE = 1.80 ka simple meaning
# RMSE roughly batata hai ki prediction errors ka magnitude ~1.8 marks hai, with larger errors getting more weight.
# kyunki RMSE large errors ko heavily penalize karta hai.
```

### 👍 Ab Linear Regression ka baseline evaluation complete hai.

### Linear Regression coefficients ko freely choose karta hai.
Kabhi-kabhi features ke coefficients unnecessarily bahut bade ho sakte hain, especially jab features ke beech correlation ho. Isse model overfit kar sakta hai.
Ridge ka idea:
Linear Regression + penalty on large coefficients

Mathematically:
\[
Loss = MSE + \alpha \sum_{j=1}^{n} \beta_j^2
\]
Yahan:
- MSE → prediction error
- β → model coefficients
- α → regularization strength
Simple analogy
Linear Regression bolta hai:
"Mujhe training data fit karna hai."

Ridge bolta hai:
"Fit karo, lekin coefficients ko unnecessarily bada mat banana."

Is technique ko L2 Regularization kehte hain.

## Ridge ( L2 Regularization)

```python
from sklearn.linear_model import Ridge
ridge_model = Ridge(alpha=1.0)
```

```python
ridge_model.fit(x_train_final, y_train)
```

```python
ridge_y_pred = ridge_model.predict(x_test_final)
```

```python
from sklearn.metrics import mean_absolute_error, r2_score, root_mean_squared_error
ridge_mae = mean_absolute_error(y_test, ridge_y_pred)
ridge_r2 = r2_score(y_test, ridge_y_pred)
ridge_rmse = root_mean_squared_error(y_test, ridge_y_pred)

print("Ridge MAE", ridge_mae)
print("Ridge R2", ridge_r2)
print("Ridge RMSE", ridge_rmse)
```

```python
ridge_001 = Ridge(alpha=100)
# I calculated by alpha = 1,10,100 - boht tiny improvements the

ridge_001.fit(x_train_final, y_train)

pred_001 = ridge_001.predict(x_test_final)
```

Here we did Hyperparameter tuning (Uppar and neeche wale code mein)

```python
mae_001 = mean_absolute_error(y_test, pred_001)
r2_001 = r2_score(y_test, pred_001)
rmse_001 = root_mean_squared_error(y_test, pred_001)

print("Alpha 0.01")
print("MAE:", mae_001)
print("R²:", r2_001)
print("RMSE:", rmse_001)
```

### Lasso Regression (L1)
Lasso uses L1 regularization, jo kuch coefficients ko exactly zero tak push kar sakti hai.

```python
from sklearn.linear_model import Lasso
lasso_model = Lasso(alpha = 0.01)
```

```python
lasso_model.fit(x_train_final, y_train)
```

```python
lasso_y_pred = lasso_model.predict(x_test_final)
```

### Evaluation

```python
lasso_mae = mean_absolute_error(y_test, lasso_y_pred)
lasso_rmse = root_mean_squared_error(y_test, lasso_y_pred)
lasso_r2 = r2_score(y_test, lasso_y_pred)

print("Lasso MAE:", lasso_mae)
print("Lasso R²:", lasso_r2)
print("Lasso RMSE:", lasso_rmse)
```
